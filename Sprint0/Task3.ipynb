<h2 style="color:DodgerBlue">Описание проекта:</h2>

Описание задачи:

Создать базовый класс PaymentMethod в C#, который будет представлять различные способы оплаты. На основе этого класса разработать 2–3 производных класса, демонстрирующих принципы наследования и полиморфизма. В каждом из классов должны быть реализованы новые атрибуты и методы, а также переопределены некоторые методы базового класса для демонстрации полиморфизма.

Требования к базовому классу PaymentMethod:

Атрибуты: ID способа оплаты (PaymentMethodId), Название способа оплаты (PaymentMethodName), Минимальная сумма (MinAmount).

Методы:

ProcessPayment(decimal amount) — метод для обработки платежа указанной суммы.

CheckMinimumAmount(decimal amount) — метод для проверки минимальной суммы платежа.

GetPaymentDetails() — метод для получения деталей способа оплаты.

Требования к производным классам:

Онлайн-оплата (OnlinePayment): Должен содержать дополнительные атрибуты, такие как URL платёжной системы (PaymentUrl). Метод ProcessPayment() должен быть переопределён для включения URL платёжной системы в процесс оплаты.

Банковский перевод (BankTransfer): Должен содержать дополнительные атрибуты, такие как Банковские данные (BankData). Метод CheckMinimumAmount() должен быть переопределён для проверки минимальной суммы платежа с учётом банковских комиссий.

Наличные (CashPayment) (если требуется третий класс): Должен содержать дополнительные атрибуты, такие как Место выдачи наличных (CashPickupPoint). Метод GetPaymentDetails() должен быть переопределён для отображения места выдачи наличных.


<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
using System;

// Базовый класс для всех способов оплаты
public class PaymentMethod
{
    public int PaymentMethodId { get; set; }
    public string PaymentMethodName { get; set; }
    public decimal MinAmount { get; set; }

    public PaymentMethod(int id, string name, decimal minAmount)
    {
        PaymentMethodId = id;
        PaymentMethodName = name;
        MinAmount = minAmount;
    }

    // Полиморфизм , virtual - метод можно переопределить в дочерних классах
    public virtual void ProcessPayment(decimal amount)
    {
        Console.WriteLine($"Обработка платежа на сумму {amount} руб.");
    }

    public virtual bool CheckMinimumAmount(decimal amount)
    {
        return amount >= MinAmount;
    }

    public virtual void GetPaymentDetails()
    {
        Console.WriteLine($"ID: {PaymentMethodId}");
        Console.WriteLine($"Название: {PaymentMethodName}");
        Console.WriteLine($"Минимальная сумма: {MinAmount} руб.");
    }
}

// Онлайн-оплата - наследуется от PaymentMethod
public class OnlinePayment : PaymentMethod
{
    public string PaymentUrl { get; set; }

    public OnlinePayment(int id, string name, decimal minAmount, string url) 
        : base(id, name, minAmount)
    {
        PaymentUrl = url;
    }

    // override - переопределяем метод родителя
    public override void ProcessPayment(decimal amount)
    {
        Console.WriteLine($"Оплата через {PaymentUrl}");
        Console.WriteLine($"Сумма: {amount} руб.");
        Console.WriteLine("Платёж прошёл успешно!");
    }

    public override void GetPaymentDetails()
    {
        base.GetPaymentDetails();
        Console.WriteLine($"URL: {PaymentUrl}");
    }
}

// Банковский перевод - наследуется от PaymentMethod
public class BankTransfer : PaymentMethod
{
    public string BankData { get; set; }

    public BankTransfer(int id, string name, decimal minAmount, string bankData) 
        : base(id, name, minAmount)
    {
        BankData = bankData;
    }

    // Переопределяем метод проверки минимальной суммы с учетом банковских комиссий
    public override bool CheckMinimumAmount(decimal amount)
    {
        decimal commission = amount * 0.01m;
        decimal total = amount + commission;
        Console.WriteLine($"Сумма: {amount}, комиссия: {commission}, итого: {total}");
        return total >= MinAmount;
    }

    public override void ProcessPayment(decimal amount)
    {
        Console.WriteLine($"Перевод на счёт: {BankData}");
        Console.WriteLine($"Сумма перевода: {amount} руб.");
    }
}

// Наличные - наследуется от PaymentMethod
public class CashPayment : PaymentMethod
{
    public string CashPickupPoint { get; set; }

    public CashPayment(int id, string name, decimal minAmount, string point) 
        : base(id, name, minAmount)
    {
        CashPickupPoint = point;
    }

    public override void GetPaymentDetails()
    {
        base.GetPaymentDetails();
        Console.WriteLine($"Место выдачи: {CashPickupPoint}");
    }

    public override void ProcessPayment(decimal amount)
    {
        Console.WriteLine($"Оплата наличными: {amount} руб.");
        Console.WriteLine($"Получить в: {CashPickupPoint}");
    }
}

// Создаём объекты
OnlinePayment onlinePay = new OnlinePayment(1, "Онлайн-оплата", 100, "https://pay.example.com");
BankTransfer bankPay = new BankTransfer(2, "Банковский перевод", 500, "Счёт №1234567890");
CashPayment cashPay = new CashPayment(3, "Наличные", 50, "Касса №5");

// Работа онлайн-оплаты
Console.WriteLine("Онлайн-оплата:");
onlinePay.GetPaymentDetails();
onlinePay.ProcessPayment(250);
Console.WriteLine($"Проверка минимальной суммы: {onlinePay.CheckMinimumAmount(250)}");

// Работа банковского перевода
Console.WriteLine("\nБанковский перевод:");
bankPay.GetPaymentDetails();
bankPay.ProcessPayment(400);
Console.WriteLine($"Проверка минимальной суммы: {bankPay.CheckMinimumAmount(400)}");

// Работа наличных
Console.WriteLine("\nНаличные:");
cashPay.GetPaymentDetails();
cashPay.ProcessPayment(300);
Console.WriteLine($"Проверка минимальной суммы: {cashPay.CheckMinimumAmount(300)}");
```
